```python
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(x, y)
plt.title("Bieu do duong")
plt.xlabel("Truc X")
plt.ylabel("Truc Y")
plt.show()
```

```python
labels = ["A", "B", "C", "D"]
values = [10, 25, 7, 18]

plt.bar(labels, values)
plt.title("Bieu do cot")
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [5, 7, 4, 8, 6]

plt.scatter(x, y)
plt.title("Scatter plot")
plt.show()
```

```python
import random

data = [random.randint(1, 100) for _ in range(1000)]

plt.hist(data, bins=20)
plt.title("Histogram")
plt.show()
```

```python
import cv2

img = cv2.imread("input.jpg")
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img_rgb)
plt.axis("off")
plt.show()
```

```python
plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.plot([1,2,3], [1,4,9])
plt.title("Line")

plt.subplot(1,2,2)
plt.bar(["A","B","C"], [5,3,7])
plt.title("Bar")

plt.show()
```
